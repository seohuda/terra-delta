# TerraDelta dataset preview
Inspect local reviewed pairs and labels. No downloads, model fitting or optimizer steps.


```python
from pathlib import Path
import sys
import numpy as np
from PIL import Image
from IPython.display import display
ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
from terradelta.data.dataset import ChangeDataset
MANIFEST = ROOT / "data/train_manifest.csv"  # Or outputs/smoke/manifest.csv
print("Select reviewed manifest:", MANIFEST)
```

```python
if MANIFEST.is_file():
    data = ChangeDataset(MANIFEST, return_auxiliary=True)
    mean = np.array([.485,.456,.406])
    std = np.array([.229,.224,.225])
    for sample in [data[i] for i in range(min(4,len(data)))]:
        views = [np.clip(np.rint((sample[k].permute(1,2,0).numpy()*std+mean)*255),0,255).astype(np.uint8) for k in ("pre","post")]
        mask = sample["mask"].numpy()
        colors = np.zeros((*mask.shape,3),np.uint8)
        colors[mask==1] = [240,90,50]
        colors[mask==2] = [50,190,100]
        colors[mask==-100] = [140,140,140]
        print(sample["id"], "background/building/tree", [(mask==i).sum() for i in (0,1,2)])
        display(Image.fromarray(np.concatenate([*views,colors],axis=1)))
else:
    print("No manifest yet. Create reviewed labels or use the tiny smoke fixture.")
```
